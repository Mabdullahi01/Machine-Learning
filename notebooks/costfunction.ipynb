```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
x_train = np.array([1.0, 1.7, 2.0, 2.5, 3.0, 3.2])
y_train = np.array([250, 300, 480,  430,   630, 730,])
```

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost_sum = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost_sum += (f_wb - y[i]) ** 2
    total_cost = (1 / (2 * m)) * cost_sum
    return total_cost
```

```python
compute_cost(x_train, y_train, 200, 100)
```

```python
w = 200
b = 100
```

```python
b = 100
w_values = np.linspace(0, 400, 100)
cost_values = np.array([compute_cost(x_train, y_train, w_value, b) for w_value in w_values])

plt.plot(w_values, cost_values, c='b', label='Cost J(w, b=100)')
plt.scatter([w], [compute_cost(x_train, y_train, w, b)], c='r', label=f'Current w = {w}')
plt.title('Cost function for linear regression')
plt.xlabel('w (slope)')
plt.ylabel('Cost J(w, b)')
plt.legend()
plt.grid(True)
plt.show()
```
